```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("/Users/xubeibei/Downloads/ds_salaries.csv")

# Preview
df.head()
```

```python
# Keep only columns needed
df = df[['job_title', 'salary_in_usd', 'employee_residence']]

# Drop any missing values
df = df.dropna(subset=['salary_in_usd'])

df.info()
```

```python
# Average salary by job title
avg_salary_by_job = (
    df.groupby('job_title')['salary_in_usd']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

# Bar chart
avg_salary_by_job.plot(kind='bar', color='skyblue')
plt.title('Top 10 Highest-Paying Data Roles')
plt.xlabel('Job Title')
plt.ylabel('Average Salary (USD)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

avg_salary_by_job
```

```python
# Average salary by location
avg_salary_by_loc = (
    df.groupby('employee_residence')['salary_in_usd']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

avg_salary_by_loc.plot(kind='bar', color='orange')
plt.title('Top 10 Locations by Average Salary (USD)')
plt.xlabel('Country')
plt.ylabel('Average Salary (USD)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

avg_salary_by_loc
```

## Findings: Skills and Locations Associated with Higher Salaries

**Skills (Job Titles):**
- Machine Learning Engineers and Data Engineers earn the highest salaries.
- Technical specializations like AI, deep learning, and data infrastructure bring higher pay.
- Data Analysts and Business Intelligence roles tend to have lower averages.

**Locations:**
- The United States, Switzerland, and Australia show the highest average salaries.
- Countries with lower costs of living (e.g., India, Brazil) report lower salary averages.
- Remote work is increasing opportunities globally but still shows regional pay gaps.
